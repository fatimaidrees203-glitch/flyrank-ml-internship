## 3. Quick look at the data (2-3 real numbers)

***Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.***

I will inspect the anonymized starter dataset to check its size, client coverage, observed decline rate, and the number of pages that meet a simple review-candidate rule. The initial checks help establish whether there is enough data and a practical prioritization problem to investigate. I will use the values printed by the executed code cell rather than manually inventing results. The decline label is based on observed recent impression trends, so it is a proxy for review priority rather than proof that a page needs a refresh.

```python

import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "fatimaidrees203-glitch/flyrank-ml-internship/"
    "main/data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(url)

# Check the size and coverage of the dataset
print("Number of pages:", len(df))
print("Number of clients:", df["client_id"].nunique())

# Measure the observed decline rate
declining = df["trend_direction"].eq("down")
print("Pages with declining impressions:", int(declining.sum()))
print("Declining rate (%):", round(declining.mean() * 100, 2))

# A transparent rule for possible review candidates:
# visible pages, reasonable search position, and low CTR
candidates = (
    (df["impressions_90d"] >= 500)
    & (df["avg_position"] > 0)
    & (df["avg_position"] <= 20)
    & (df["ctr"] < 0.5)
)

print("Pages matching review-candidate rule:", int(candidates.sum()))
print("Candidate share (%):", round(candidates.mean() * 100, 2))

```

### Findings from the Starter Dataset

The starter dataset contains **30,000 pages across 32 anonymized clients**. Out of these pages, **16,262 (54.21%)** have declining impressions. This suggests that investigating content performance and review prioritization may be useful.

A simple rule based on impressions, average search position, and CTR identified **9,759 pages (32.53%)** as potential review candidates. This is an initial screening rule, not a validated recommendation system.

These results support exploring the Refresh / Content Opportunity Scoring lane. The next step is to evaluate whether a ranking approach can prioritize useful pages more effectively than a simple baseline. Declining impressions alone do not prove that a page needs refreshing, and the analysis cannot establish that refreshing a page will improve performance.
